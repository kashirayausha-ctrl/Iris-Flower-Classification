```python
# ================================================================
# 🌸 IRIS FLOWER CLASSIFICATION WITH MACHINE LEARNING
# ================================================================
# Project: Iris Flower Classification
# Model: Logistic Regression
# Dataset: Iris Dataset
#
# Workflow:
# 1. Library Imports
# 2. Project and Data Path
# 3. Dataset Loading
# 4. Dataset Preview
# 5. Structural Audit
# 6. Missing Value Audit
# 7. Duplicate Record Audit
# 8. Categorical Audit
# 9. Numerical Audit
# 10. Data Validation
# 11. Data Cleaning
# 12. Exploratory Data Analysis
# 13. Feature Relationship Analysis
# 14. Correlation Analysis
# 15. Train-Test Split
# 16. Feature Scaling
# 17. Logistic Regression
# 18. Model Evaluation
# 19. Confusion Matrix
# 20. Feature Coefficients
# 21. New Flower Prediction
# 22. Final Findings
# ================================================================


# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from IPython.display import display


# Display configuration
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 120)
pd.set_option("display.float_format", lambda x: f"{x:,.3f}")

# Plotting configuration
sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")


# ================================================================
# 2. PROJECT AND DATA PATH
# ================================================================

# Change this path if your Iris.csv is stored somewhere else
FILE_PATH = Path(r"C:\Users\User\Desktop\Iris.csv")

if not FILE_PATH.exists():
    raise FileNotFoundError(
        f"Iris dataset not found at:\n{FILE_PATH}\n\n"
        "Please update FILE_PATH to the correct location of Iris.csv."
    )

print("\n" + "=" * 80)
print("PROJECT AND DATA PATH")
print("=" * 80)

print("\nDataset path:")
print(FILE_PATH)


# ================================================================
# 3. LOAD DATASET
# ================================================================

df = pd.read_csv(FILE_PATH)

print("\nDataset loaded successfully.")
print(f"Dataset shape: {df.shape}")


# ================================================================
# 4. DATASET PREVIEW
# ================================================================

print("\n" + "=" * 80)
print("FIRST 5 ROWS OF THE DATASET")
print("=" * 80)

display(df.head())


# ================================================================
# 5. STRUCTURAL AUDIT
# ================================================================

print("\n" + "=" * 80)
print("DATASET STRUCTURAL AUDIT")
print("=" * 80)

print(f"\nRows:    {df.shape[0]:,}")
print(f"Columns: {df.shape[1]:,}")

print("\nColumn names:")

for i, column in enumerate(df.columns, start=1):
    print(f"{i:2}. {column}")


print("\nData types:")

display(
    df.dtypes.to_frame(name="dtype")
)


print("\nNon-null counts:")

display(
    df.notna().sum().to_frame(name="non_null_count")
)


print("\nMemory usage:")

print(
    f"{df.memory_usage(deep=True).sum() / 1024**2:.2f} MB"
)


# ================================================================
# 6. MISSING VALUE AUDIT
# ================================================================

print("\n" + "=" * 80)
print("MISSING VALUE AUDIT")
print("=" * 80)

missing_audit = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percentage": df.isna().mean() * 100,
    "non_missing_count": df.notna().sum()
})

display(missing_audit)


# ================================================================
# 7. DUPLICATE RECORD AUDIT
# ================================================================

print("\n" + "=" * 80)
print("DUPLICATE RECORD AUDIT")
print("=" * 80)

duplicate_count = df.duplicated().sum()
duplicate_rows = df.duplicated(keep=False).sum()

print(
    f"\nExact duplicate rows excluding first occurrence: "
    f"{duplicate_count:,}"
)

print(
    f"Rows belonging to duplicate groups: "
    f"{duplicate_rows:,}"
)

if duplicate_rows > 0:

    print("\nSample duplicate records:")

    display(
        df.loc[df.duplicated(keep=False)]
        .sort_values(list(df.columns))
        .head(20)
    )

else:

    print("\nNo exact duplicate rows detected.")


# ================================================================
# 8. CATEGORICAL AUDIT
# ================================================================
# Updated for newer Pandas versions.
# Including "string" prevents the select_dtypes warning.
# ================================================================

print("\n" + "=" * 80)
print("CATEGORICAL AUDIT")
print("=" * 80)

categorical_columns = df.select_dtypes(
    include=["object", "string", "category"]
).columns

if len(categorical_columns) == 0:

    print("\nNo categorical columns detected.")

else:

    for column in categorical_columns:

        print(f"\n--- {column} ---")

        print(
            f"Unique values: "
            f"{df[column].nunique(dropna=False)}"
        )

        value_counts = df[column].value_counts(
            dropna=False
        )

        display(
            value_counts.to_frame(name="count")
        )


# ================================================================
# 9. NUMERICAL FEATURE AUDIT
# ================================================================

print("\n" + "=" * 80)
print("NUMERICAL FEATURE AUDIT")
print("=" * 80)

numeric_columns = df.select_dtypes(
    include="number"
).columns

numerical_summary = df[numeric_columns].describe().T

numerical_summary["missing"] = (
    df[numeric_columns].isna().sum()
)

numerical_summary["unique"] = (
    df[numeric_columns].nunique()
)

display(numerical_summary)


# ================================================================
# 10. DATA VALIDATION
# ================================================================

print("\n" + "=" * 80)
print("DATA VALIDATION")
print("=" * 80)


# Features used for classification
feature_columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

# Target variable
target_column = "Species"


# ------------------------------------------------
# Check required columns
# ------------------------------------------------

required_columns = feature_columns + [
    target_column
]

missing_required_columns = [
    column
    for column in required_columns
    if column not in df.columns
]

if missing_required_columns:

    raise ValueError(
        f"Required columns are missing: "
        f"{missing_required_columns}"
    )

print("\nRequired columns found successfully.")


# ------------------------------------------------
# Check missing values
# ------------------------------------------------

important_missing = (
    df[required_columns]
    .isna()
    .sum()
)

print("\nMissing values in important columns:")

display(
    important_missing.to_frame(
        name="missing_count"
    )
)


# ------------------------------------------------
# Check negative values
# ------------------------------------------------

negative_values = (
    df[feature_columns] < 0
).sum()

print("\nNegative values in measurement features:")

display(
    negative_values.to_frame(
        name="negative_count"
    )
)


# ------------------------------------------------
# Check target classes
# ------------------------------------------------

print("\nTarget classes:")

print(
    df[target_column].dropna().unique()
)


# ================================================================
# 11. DATA CLEANING
# ================================================================

print("\n" + "=" * 80)
print("DATA CLEANING")
print("=" * 80)

# Create a copy so that the original data remains unchanged
df_clean = df.copy()

initial_rows = len(df_clean)


# ------------------------------------------------
# Remove completely empty rows
# ------------------------------------------------

df_clean = df_clean.dropna(
    how="all"
)

after_empty_rows = len(df_clean)


# ------------------------------------------------
# Remove exact duplicate rows
# ------------------------------------------------

df_clean = df_clean.drop_duplicates()

after_duplicates = len(df_clean)


# ------------------------------------------------
# Remove rows missing important values
# ------------------------------------------------

df_clean = df_clean.dropna(
    subset=required_columns
)

final_rows = len(df_clean)


print(
    f"\nInitial rows: "
    f"{initial_rows:,}"
)

print(
    f"After removing empty rows: "
    f"{after_empty_rows:,}"
)

print(
    f"After removing duplicates: "
    f"{after_duplicates:,}"
)

print(
    f"Final cleaned rows: "
    f"{final_rows:,}"
)

print(
    f"\nTotal rows removed during cleaning: "
    f"{initial_rows - final_rows:,}"
)


# ================================================================
# 12. CLEANED DATA VALIDATION
# ================================================================

print("\n" + "=" * 80)
print("CLEANED DATA VALIDATION")
print("=" * 80)

print("\nMissing values after cleaning:")

display(
    df_clean[required_columns]
    .isna()
    .sum()
    .to_frame(name="missing_count")
)


print("\nDuplicate rows after cleaning:")

print(
    df_clean.duplicated().sum()
)


print("\nCleaned dataset shape:")

print(
    df_clean.shape
)


# ================================================================
# 13. EXPLORATORY DATA ANALYSIS
# ================================================================

print("\n" + "=" * 80)
print("EXPLORATORY DATA ANALYSIS")
print("=" * 80)


# ------------------------------------------------
# 13.1 Species Distribution
# ------------------------------------------------

print("\nSpecies distribution:")

species_counts = (
    df_clean[target_column]
    .value_counts()
)

display(
    species_counts.to_frame(
        name="count"
    )
)


plt.figure(figsize=(7, 5))

sns.countplot(
    data=df_clean,
    x=target_column
)

plt.title("Iris Species Distribution")
plt.xlabel("Species")
plt.ylabel("Number of Samples")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 13.2 Feature Distributions
# ------------------------------------------------

for feature in feature_columns:

    plt.figure(figsize=(7, 5))

    sns.histplot(
        data=df_clean,
        x=feature,
        hue=target_column,
        kde=True
    )

    plt.title(
        f"Distribution of {feature}"
    )

    plt.xlabel(feature)
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()


# ------------------------------------------------
# 13.3 Boxplots
# ------------------------------------------------

for feature in feature_columns:

    plt.figure(figsize=(7, 5))

    sns.boxplot(
        data=df_clean,
        x=target_column,
        y=feature
    )

    plt.title(
        f"{feature} by Species"
    )

    plt.xlabel("Species")
    plt.ylabel(feature)

    plt.tight_layout()
    plt.show()


# ================================================================
# 14. FEATURE RELATIONSHIP ANALYSIS
# ================================================================

print("\n" + "=" * 80)
print("FEATURE RELATIONSHIP ANALYSIS")
print("=" * 80)

sns.pairplot(
    df_clean[
        feature_columns + [target_column]
    ],
    hue=target_column,
    diag_kind="hist"
)

plt.suptitle(
    "Pairwise Relationships Between Iris Features",
    y=1.02
)

plt.show()


# ================================================================
# 15. CORRELATION ANALYSIS
# ================================================================

print("\n" + "=" * 80)
print("FEATURE CORRELATION ANALYSIS")
print("=" * 80)

correlation_matrix = (
    df_clean[feature_columns]
    .corr()
)

display(correlation_matrix)


plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    square=True
)

plt.title(
    "Correlation Matrix of Iris Features"
)

plt.tight_layout()
plt.show()


# ================================================================
# 16. PREPARE FEATURES AND TARGET
# ================================================================

print("\n" + "=" * 80)
print("FEATURE AND TARGET PREPARATION")
print("=" * 80)

X = df_clean[
    feature_columns
]

y = df_clean[
    target_column
]

print("\nFeatures used:")

for feature in feature_columns:
    print(f" - {feature}")

print("\nTarget variable:")
print(f" - {target_column}")


# ================================================================
# 17. TRAIN-TEST SPLIT
# ================================================================

print("\n" + "=" * 80)
print("TRAIN-TEST SPLIT")
print("=" * 80)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(
    f"\nTraining samples: "
    f"{len(X_train):,}"
)

print(
    f"Testing samples: "
    f"{len(X_test):,}"
)

print(
    f"\nTraining percentage: "
    f"{len(X_train) / len(X) * 100:.1f}%"
)

print(
    f"Testing percentage: "
    f"{len(X_test) / len(X) * 100:.1f}%"
)


# ================================================================
# 18. FEATURE SCALING
# ================================================================

print("\n" + "=" * 80)
print("FEATURE SCALING")
print("=" * 80)

scaler = StandardScaler()

# Fit only on training data
X_train_scaled = scaler.fit_transform(
    X_train
)

# Apply the same scaler to test data
X_test_scaled = scaler.transform(
    X_test
)

print(
    "\nFeatures standardized successfully."
)

print(
    "\nTraining feature means after scaling:"
)

print(
    X_train_scaled.mean(axis=0)
)

print(
    "\nTraining feature standard deviations after scaling:"
)

print(
    X_train_scaled.std(axis=0)
)


# ================================================================
# 19. TRAIN LOGISTIC REGRESSION MODEL
# ================================================================

print("\n" + "=" * 80)
print("MODEL TRAINING")
print("=" * 80)

model = LogisticRegression(
    max_iter=200,
    random_state=42
)

model.fit(
    X_train_scaled,
    y_train
)

print(
    "\nLogistic Regression model "
    "trained successfully."
)


# ================================================================
# 20. MAKE PREDICTIONS
# ================================================================

y_pred = model.predict(
    X_test_scaled
)

print(
    "\nPredictions generated successfully."
)


# ================================================================
# 21. MODEL ACCURACY
# ================================================================

print("\n" + "=" * 80)
print("MODEL ACCURACY")
print("=" * 80)

accuracy = accuracy_score(
    y_test,
    y_pred
)

print(
    f"\nAccuracy: "
    f"{accuracy:.4f}"
)

print(
    f"Accuracy Percentage: "
    f"{accuracy * 100:.2f}%"
)


# ================================================================
# 22. CLASSIFICATION REPORT
# ================================================================

print("\n" + "=" * 80)
print("CLASSIFICATION REPORT")
print("=" * 80)

report = classification_report(
    y_test,
    y_pred
)

print(report)


# ================================================================
# 23. CONFUSION MATRIX
# ================================================================

print("\n" + "=" * 80)
print("CONFUSION MATRIX")
print("=" * 80)

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\nConfusion Matrix:")
print(cm)


plt.figure(figsize=(7, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=model.classes_,
    yticklabels=model.classes_
)

plt.title(
    "Confusion Matrix - Logistic Regression"
)

plt.xlabel(
    "Predicted Label"
)

plt.ylabel(
    "Actual Label"
)

plt.tight_layout()
plt.show()


# ================================================================
# 24. MODEL COEFFICIENTS
# ================================================================

print("\n" + "=" * 80)
print("MODEL FEATURE COEFFICIENTS")
print("=" * 80)

coefficient_df = pd.DataFrame(
    model.coef_,
    columns=feature_columns,
    index=model.classes_
)

display(
    coefficient_df
)


plt.figure(figsize=(10, 6))

coefficient_df.T.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title(
    "Logistic Regression Feature Coefficients"
)

plt.xlabel(
    "Features"
)

plt.ylabel(
    "Coefficient Value"
)

plt.xticks(
    rotation=45
)

plt.tight_layout()
plt.show()


# ================================================================
# 25. NEW FLOWER PREDICTION
# ================================================================

print("\n" + "=" * 80)
print("NEW FLOWER PREDICTION")
print("=" * 80)


# Example measurements:
# Sepal Length = 5.1 cm
# Sepal Width  = 3.5 cm
# Petal Length = 1.4 cm
# Petal Width  = 0.2 cm

new_flower = pd.DataFrame(
    [
        [5.1, 3.5, 1.4, 0.2]
    ],
    columns=feature_columns
)

print("\nNew flower measurements:")

display(
    new_flower
)


# Scale the new flower using
# the same scaler used during training

new_flower_scaled = scaler.transform(
    new_flower
)


# Predict species

prediction = model.predict(
    new_flower_scaled
)


# Prediction probabilities

prediction_probabilities = (
    model.predict_proba(
        new_flower_scaled
    )
)


predicted_species = prediction[0]


print(
    "\nPredicted Species:"
)

print(
    predicted_species
)


print(
    "\nPrediction Probabilities:"
)

probability_df = pd.DataFrame(
    prediction_probabilities,
    columns=model.classes_
)

display(
    probability_df
)


# ================================================================
# 26. FINAL MODEL SUMMARY
# ================================================================

print("\n" + "=" * 80)
print("FINAL MODEL SUMMARY")
print("=" * 80)

print(
    f"""
Model:
- Logistic Regression

Dataset:
- Original observations: {len(df):,}
- Cleaned observations:  {len(df_clean):,}

Features:
- Sepal Length
- Sepal Width
- Petal Length
- Petal Width

Target:
- Species

Train-Test Split:
- Training data: 80%
- Testing data: 20%

Evaluation:
- Accuracy: {accuracy * 100:.2f}%

Example New Flower:
- Sepal Length: 5.1 cm
- Sepal Width: 3.5 cm
- Petal Length: 1.4 cm
- Petal Width: 0.2 cm

Predicted Species:
- {predicted_species}
"""
)


# ================================================================
# 27. FINAL DATA SCIENCE FINDINGS
# ================================================================

print("\n" + "=" * 80)
print("FINAL FINDINGS")
print("=" * 80)

print(
    f"""
1. The dataset contains measurements of Iris flowers
   belonging to different species.

2. Four numerical measurements were used as input features:
   Sepal Length, Sepal Width, Petal Length and Petal Width.

3. The target variable is Species.

4. The dataset was audited for structure, missing values,
   duplicate records, categorical values and numerical values.

5. Data cleaning was performed only after the initial audit.

6. The numerical features were standardized using
   StandardScaler.

7. Logistic Regression was trained using 80% of the dataset
   and evaluated using the remaining 20%.

8. The model achieved an accuracy of:
   {accuracy * 100:.2f}%

9. Model performance was evaluated using accuracy,
   precision, recall, F1-score and a confusion matrix.

10. Feature relationships were explored using histograms,
    boxplots, pairplots and a correlation heatmap.

11. A new Iris flower was provided to the trained model.

12. The predicted species for the example flower is:
    {predicted_species}
"""
)


# ================================================================
# 28. PROJECT COMPLETE
# ================================================================

print("\n" + "=" * 80)
print("✅ IRIS FLOWER CLASSIFICATION PROJECT COMPLETE")
print("=" * 80)
```
